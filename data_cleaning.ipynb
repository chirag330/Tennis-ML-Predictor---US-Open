## Building Primitive Model

```python
# Create our modeling population
completed_matches = df[
    (df["status"] == "FINISHED") &
    (df["status_extra"] == "FINISHED") &
    (df["winner_code"].notna())
].copy()
```

```python
completed_matches.size
```

```python
completed_matches["home_win"] = (
    completed_matches["winner_code"] == 1
).astype(int)
```

```python
completed_matches["home_win"].value_counts()
```

## Building baseline

```python
baseline = completed_matches.dropna(
    subset=["home_rank", "away_rank"]
).copy()
```

```python
baseline.head(10)
```

```python
baseline["predicted_home_win"] = (
    baseline["home_rank"] < baseline["away_rank"]
).astype(int)
```

```python
baseline.head(10)
```

```python
from sklearn.metrics import accuracy_score

baseline_accuracy = accuracy_score(
    baseline["home_win"],
    baseline["predicted_home_win"]
)

print(f"The baseline accuracy score is: {baseline_accuracy*100}%")
```

## Logistic Regression Model V1

```python
model_data = completed_matches.copy()

model_data["rank_diff"] = (
    model_data["away_rank"] - model_data["home_rank"]
)

model_data["points_diff"] = (
    model_data["home_points"] - model_data["away_points"]
)
```

```python
model_data[
    [
        "home_name",
        "away_name",
        "home_rank",
        "away_rank",
        "rank_diff",
        "home_points",
        "away_points",
        "points_diff",
        "home_win"
    ]
].head(50)
```

```python
model_data[
    ["home_rank", "away_rank", "home_points", "away_points"]
].isna().mean() * 100
```

```python
v1_data = model_data.dropna(
    subset=[
        "home_rank",
        "away_rank",
        "home_points",
        "away_points"
    ]
).copy()
```

```python
print("Original completed matches:", len(model_data))
print("V1 usable matches:", len(v1_data))
print(
    "Percentage retained:",
    len(v1_data) / len(model_data) * 100
)
```

```python
# Splitting the data
train_v1 = v1_data[
    v1_data["season_year"] <= 2024
].copy()

validation_v1 = v1_data[
    v1_data["season_year"] == 2025
].copy()
```

```python
train_v1.head(10)
```

```python
# Checking the size of the training and validation data sets
print("Training data: ", len(train_v1))
print("Validation data: ", len(validation_v1))
```

```python
print(train_v1["season_year"].value_counts().sort_index())
print()
print(validation_v1["season_year"].value_counts().sort_index())
```

```python
# Defining the covariates and response
features_v1 = ["rank_diff", "points_diff"]

X_train = train_v1[features_v1]
y_train = train_v1["home_win"]

X_val = validation_v1[features_v1]
y_val = validation_v1["home_win"]
```

```python
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

print("X_val shape:", X_val.shape)
print("y_val shape:", y_val.shape)
```

```python
X_train.head()
```

```python
# Fitting the logistic model
from sklearn.linear_model import LogisticRegression

log_reg_v1 = LogisticRegression(max_iter=2000)

log_reg_v1.fit(X_train, y_train)
```

```python
print("Intercept:", log_reg_v1.intercept_[0])

for feature, coefficient in zip(
    features_v1,
    log_reg_v1.coef_[0]
):
    print(feature, ":", coefficient)
```

```python
# Predictions made by our model
val_predictions = log_reg_v1.predict(X_val)
```

```python
# The prediction probabilites
val_probabilities = log_reg_v1.predict_proba(X_val)[:, 1]
```

```python
print("The value predictions for the first 10 observations: ", val_predictions[:10])
print("The value probabilities for the first 10 observations: ", val_probabilities[:10])
```

```python
# Calculating the accuracy score
from sklearn.metrics import accuracy_score

v1_accuracy = accuracy_score(
    y_val,
    val_predictions
)

print("Logistic Regression V1 Accuracy:", round(v1_accuracy*100, 2), "%")
```

```python
from sklearn.metrics import log_loss

v1_log_loss = log_loss(
    y_val,
    val_probabilities
)

print("Logistic Regression V1 Log Loss:", v1_log_loss)
```

## Logistic Regression V2
- This model will use the player's previous performance as the covariates
- The performance in this case would be the average of the ace points, win rates, difference in the games within winning sets, average of points lost due to double faults.



```python
# Creating the history for the home player
home_history = completed_matches[[
    "match_id",
    "date_timestamp",
    "season_year",
    "surface",
    "home_id",
    "home_name",
    "away_id",
    "away_name",
    "winner_code",
    "home_aces",
    "home_double_faults",
    "home_service_points_won_perc",
    "home_return_points_won_perc",
    "home_break_points_won_perc",
    "home_break_points_saved_perc"
]].copy()
```

```python
home_history = home_history.rename(columns={
    "home_id": "player_id",
    "home_name": "player_name",
    "away_id": "opponent_id",
    "away_name": "opponent_name",
    "home_aces": "aces",
    "home_double_faults": "double_faults",
    "home_service_points_won_perc": "service_points_won_perc",
    "home_return_points_won_perc": "return_points_won_perc",
    "home_break_points_won_perc": "break_points_won_perc",
    "home_break_points_saved_perc": "break_points_saved_perc"
})
```

```python
home_history["won"] = (
    home_history["winner_code"] == 1
).astype(int)
```

```python
home_history.head(50)
```

```python
# Creating history for away players
away_history = completed_matches[[
    "match_id",
    "date_timestamp",
    "season_year",
    "surface",
    "home_id",
    "home_name",
    "away_id",
    "away_name",
    "winner_code",
    "home_aces",
    "home_double_faults",
    "home_service_points_won_perc",
    "home_return_points_won_perc",
    "home_break_points_won_perc",
    "home_break_points_saved_perc"
]].copy()
```

```python
away_history = away_history.rename(columns={
    "home_id": "player_id",
    "home_name": "player_name",
    "away_id": "opponent_id",
    "away_name": "opponent_name",
    "home_aces": "aces",
    "home_double_faults": "double_faults",
    "home_service_points_won_perc": "service_points_won_perc",
    "home_return_points_won_perc": "return_points_won_perc",
    "home_break_points_won_perc": "break_points_won_perc",
    "home_break_points_saved_perc": "break_points_saved_perc"
})
```

```python
away_history["won"] = (
    away_history["winner_code"] == 2
).astype(int)
```

```python
away_history.head(50)
```

```python
player_history = pd.concat(
    [home_history, away_history],
    ignore_index=True
)
```

```python
player_history = player_history.sort_values(
    ["date_timestamp", "match_id"]
).reset_index(drop=True)
```
