## Logistic Regression Model V1

```python
model_data = completed_matches.copy()

model_data["rank_diff"] = (
    model_data["away_rank"] - model_data["home_rank"]
)

model_data["points_diff"] = (
    model_data["home_points"] - model_data["away_points"]
)
```

```python
v1_data = model_data.dropna(
    subset=[
        "home_rank",
        "away_rank",
        "home_points",
        "away_points"
    ]
).copy()
```

```python
print("Original completed matches:", len(model_data))
print("V1 usable matches:", len(v1_data))
print(
    "Percentage retained:",
    len(v1_data) / len(model_data) * 100
)
```

```python
# Splitting the data
train_v1 = v1_data[
    v1_data["season_year"] <= 2024
].copy()

validation_v1 = v1_data[
    v1_data["season_year"] == 2025
].copy()
```

```python
train_v1.head(10)
```

```python
# Defining the covariates and response
features_v1 = ["rank_diff", "points_diff"]

X_train = train_v1[features_v1]
y_train = train_v1["home_win"]

X_val = validation_v1[features_v1]
y_val = validation_v1["home_win"]
```

```python
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

print("X_val shape:", X_val.shape)
print("y_val shape:", y_val.shape)
```

```python
# Fitting the logistic model
from sklearn.linear_model import LogisticRegression

log_reg_v1 = LogisticRegression(max_iter=2000)

log_reg_v1.fit(X_train, y_train)
```

```python
print("Intercept:", log_reg_v1.intercept_[0])

for feature, coefficient in zip(
    features_v1,
    log_reg_v1.coef_[0]
):
    print(feature, ":", coefficient)
```

```python
# Predictions made by our model
val_predictions = log_reg_v1.predict(X_val)
```

```python
# The prediction probabilites
val_probabilities = log_reg_v1.predict_proba(X_val)[:, 1]
```

```python
print("The value predictions for the first 10 observations: ", val_predictions[:10])
print("The value probabilities for the first 10 observations: ", val_probabilities[:10])
```

```python
# Calculating the accuracy score
from sklearn.metrics import accuracy_score

v1_accuracy = accuracy_score(
    y_val,
    val_predictions
)

print("Logistic Regression V1 Accuracy:", round(v1_accuracy*100, 2), "%")
```

```python
away_history = away_history.rename(columns={
    "away_id": "player_id",
    "away_name": "player_name",
    "home_id": "opponent_id",
    "home_name": "opponent_name",
    "away_aces": "aces",
    "away_double_faults": "double_faults",
    "away_service_points_won_perc": "service_points_won_perc",
    "away_return_points_won_perc": "return_points_won_perc",
    "away_break_points_won_perc": "break_points_won_perc",
    "away_break_points_saved_perc": "break_points_saved_perc"
})
```

```python
# Rolling feature: Win rate of the last 10 matches
player_history["win_rate_last_10"] = (
    player_history
    .groupby("player_id")["won"]
    .transform(
        lambda s: s.shift(1).rolling(10, min_periods=1).mean()
    )
)
```

```python
# Rolling feature: Average number of aces of the last 10 matches
player_history["avg_aces_last_10"] = (
    player_history
    .groupby("player_id")["aces"]
    .transform(
        lambda s: s.shift(1).rolling(10, min_periods=1).mean()
    )
)
```

```python
player_history.tail(50)
```

```python
# Rolling feature: Average number of double faults in the last 10 matches
player_history["avg_double_faults_last_10"] = (
    player_history
    .groupby("player_id")["double_faults"]
    .transform(
        lambda s: s.shift(1).rolling(10, min_periods=1).mean()
    )
)
```

```python
# Rolling feature: Average service points won % in the last 10 matches
player_history["avg_service_points_won_last_10"] = (
    player_history
    .groupby("player_id")["service_points_won_perc"]
    .transform(
        lambda s: s.shift(1).rolling(10, min_periods=1).mean()
    )
)
```

```python
player_history[["player_name", "won", "win_rate_last_10", "avg_aces_last_10","avg_double_faults_last_10", "avg_service_points_won_last_10"]].tail(20)
```

```python
# Rolling feature: Average return points won % over the last 10 matches
player_history["avg_return_points_won_last_10"] = (
    player_history
    .groupby("player_id")["return_points_won_perc"]
    .transform(
        lambda s: s.shift(1).rolling(10, min_periods=1).mean()
    )
)
```

```python
# Rolling feature: Average break points won %
player_history["avg_break_points_won_last_10"] = (
    round(player_history
    .groupby("player_id")["break_points_won_perc"]
    .transform(
        lambda s: s.shift(1).rolling(10, min_periods=1).mean()
    ),2)
)
```

```python
# Rolling feature: Average Break Points saved%
player_history["avg_break_points_saved_last_10"] = (
    player_history
    .groupby("player_id")["break_points_saved_perc"]
    .transform(
        lambda s: s.shift(1).rolling(10, min_periods=1).mean()
    )
)
```

```python
# Feature: Sets won by home and away players individually
set_numbers = [1, 2, 3, 4, 5]

# Calculate sets won by home player
completed_matches["home_sets_won_calc"] = 0

for i in set_numbers:
    home_col = f"home_set_{i}_score"
    away_col = f"away_set_{i}_score"

    completed_matches["home_sets_won_calc"] += (
        completed_matches[home_col] > completed_matches[away_col]
    ).astype(int)

# Calculate sets won by away player
completed_matches["away_sets_won_calc"] = 0

for i in set_numbers:
    home_col = f"home_set_{i}_score"
    away_col = f"away_set_{i}_score"

    completed_matches["away_sets_won_calc"] += (
        completed_matches[away_col] > completed_matches[home_col]
    ).astype(int)
```

```python
completed_matches[[
    "home_name",
    "away_name",
    "home_set_1_score",
    "away_set_1_score",
    "home_set_2_score",
    "away_set_2_score",
    "home_set_3_score",
    "away_set_3_score",
    "home_sets_won_calc",
    "away_sets_won_calc"
]].head(10)
```

```python
# Feature: Total games won by each player in each match
completed_matches["home_games_won"] = 0
completed_matches["away_games_won"] = 0

for i in set_numbers:
    home_col = f"home_set_{i}_score"
    away_col = f"away_set_{i}_score"

    completed_matches["home_games_won"] += (
        completed_matches[home_col].fillna(0)
    )

    completed_matches["away_games_won"] += (
        completed_matches[away_col].fillna(0)
    )
```

```python
completed_matches["home_game_diff"] = (
    completed_matches["home_games_won"]
    - completed_matches["away_games_won"]
)

completed_matches["away_game_diff"] = (
    completed_matches["away_games_won"]
    - completed_matches["home_games_won"]
)
```

```python
completed_matches["total_games"] = (
    completed_matches["home_games_won"]
    + completed_matches["away_games_won"]
)

completed_matches["home_games_won_pct"] = (
    round((completed_matches["home_games_won"]
    / completed_matches["total_games"])*100, 2)
)

completed_matches["away_games_won_pct"] = (
    round((completed_matches["away_games_won"]
    / completed_matches["total_games"])*100, 2)
)
```

```python
completed_matches[[
    "home_name",
    "away_name",
    "home_games_won",
    "away_games_won",
    "home_game_diff",
    "home_games_won_pct",
    "away_games_won_pct"
]].head(10)
```

```python
completed_matches["total_sets_played"] = (
    completed_matches["home_sets_won_calc"]
    + completed_matches["away_sets_won_calc"]
)
```

```python
home_score_stats = completed_matches[[
    "match_id",
    "home_sets_won_calc",
    "total_sets_played",
    "home_games_won_pct",
    "home_game_diff"
]].copy()

home_score_stats = home_score_stats.rename(columns={
    "home_sets_won_calc": "sets_won",
    "home_games_won_pct": "games_won_pct",
    "home_game_diff": "game_diff"
})

home_history = home_history.merge(
    home_score_stats,
    on="match_id",
    how="left"
)
```

```python
home_history[[
    "player_name",
    "sets_won",
    "total_sets_played",
    "games_won_pct",
    "game_diff"
]].head(10)
```

```python
away_score_stats = completed_matches[[
    "match_id",
    "away_sets_won_calc",
    "total_sets_played",
    "away_games_won_pct",
    "away_game_diff"
]].copy()

away_score_stats = away_score_stats.rename(columns={
    "away_sets_won_calc": "sets_won",
    "away_games_won_pct": "games_won_pct",
    "away_game_diff": "game_diff"
})

away_history = away_history.merge(
    away_score_stats,
    on="match_id",
    how="left"
)
```

```python
away_history[[
    "player_name",
    "sets_won",
    "total_sets_played",
    "games_won_pct",
    "game_diff"
]].head(10)
```

```python
# Concatinating the two histories
player_history = pd.concat(
    [home_history, away_history],
    ignore_index=True
)

player_history = player_history.sort_values(
    ["date_timestamp", "match_id"]
).reset_index(drop=True)
```

```python
player_history["sets_won_pct_last_10"] = (
    player_history
    .groupby("player_id")
    .apply(
        lambda g: (
            g["sets_won"].shift(1).rolling(10, min_periods=1).sum()
            /
            g["total_sets_played"].shift(1).rolling(10, min_periods=1).sum()
        )
    )
    .reset_index(level=0, drop=True)
)
```

```python
player_history[[
    "player_name",
    "sets_won",
    "total_sets_played",
    "sets_won_pct_last_10"
]].tail(20)
```

```python
# Feature: Wins on hard surface
player_history["is_hard"] = (
    player_history["surface"].str.lower() == "hard"
).astype(int)
```

```python
player_history["hard_win"] = (
    player_history["won"] * player_history["is_hard"]
)
```

```python
player_history["previous_hard_wins"] = (
    player_history
    .groupby("player_id")["hard_win"]
    .transform(lambda s: s.shift(1).cumsum())
)

player_history["previous_hard_matches"] = (
    player_history
    .groupby("player_id")["is_hard"]
    .transform(lambda s: s.shift(1).cumsum())
)
```

```python
player_history["hard_win_rate"] = (
    player_history["previous_hard_wins"]
    / player_history["previous_hard_matches"]
)
```

```python
# The function for calculating the expected Elo score
def expected_score(player_rating, opponent_rating):
    return 1 / (
        1 + 10 ** ((opponent_rating - player_rating) / 400)
    )

```

```python
# Function to update Elo
def update_elo(rating, actual_score, expected_score, k=32):
    return rating + k * (actual_score - expected_score)
```

```python
player_a_rating = 1700
player_b_rating = 1500

player_a_expected = expected_score(
    player_a_rating,
    player_b_rating
)

player_b_expected = expected_score(
    player_b_rating,
    player_a_rating
)

print("A expected:", player_a_expected)
print("B expected:", player_b_expected)
```

```python
# Assume player A wins the match

new_a_rating = update_elo(
    player_a_rating,
    actual_score=1,
    expected_score=player_a_expected
)

new_b_rating = update_elo(
    player_b_rating,
    actual_score=0,
    expected_score=player_b_expected
)

print("A:", player_a_rating, "->", new_a_rating)
print("B:", player_b_rating, "->", new_b_rating)
```

```python
# Generating pre-match Elo scores
player_elos = {}
```

```python
STARTING_ELO = 1500
K = 32
```

```python
home_elo_before = []
away_elo_before = []
```

```python
# Looping through every row in elo_data and updating the elo scores accordingly
for _, row in elo_data.iterrows():

    home_id = row["home_id"]
    away_id = row["away_id"]

    home_rating = player_elos.get(home_id, STARTING_ELO)
    away_rating = player_elos.get(away_id, STARTING_ELO)

    home_elo_before.append(home_rating)
    away_elo_before.append(away_rating)

    home_expected = expected_score(
        home_rating,
        away_rating
    )

    away_expected = expected_score(
        away_rating,
        home_rating
    )

    if row["winner_code"] == 1:
        home_actual = 1
        away_actual = 0
    else:
        home_actual = 0
        away_actual = 1

    new_home_rating = update_elo(
        home_rating,
        home_actual,
        home_expected,
        k=K
    )

    new_away_rating = update_elo(
        away_rating,
        away_actual,
        away_expected,
        k=K
    )

    player_elos[home_id] = new_home_rating
    player_elos[away_id] = new_away_rating
```

```python
hard_elos = {}

home_hard_elo_before = []
away_hard_elo_before = []

for _, row in elo_data.iterrows():

    home_id = row["home_id"]
    away_id = row["away_id"]

    home_hard_rating = hard_elos.get(
        home_id,
        STARTING_ELO
    )

    away_hard_rating = hard_elos.get(
        away_id,
        STARTING_ELO
    )

    # Save PRE-MATCH hard Elo
    home_hard_elo_before.append(home_hard_rating)
    away_hard_elo_before.append(away_hard_rating)

    # Only update hard Elo after hard-court matches
    if str(row["surface"]).lower() == "hard":

        home_expected = expected_score(
            home_hard_rating,
            away_hard_rating
        )

        away_expected = expected_score(
            away_hard_rating,
            home_hard_rating
        )

        if row["winner_code"] == 1:
            home_actual = 1
            away_actual = 0
        else:
            home_actual = 0
            away_actual = 1

        hard_elos[home_id] = update_elo(
            home_hard_rating,
            home_actual,
            home_expected,
            k=K
        )

        hard_elos[away_id] = update_elo(
            away_hard_rating,
            away_actual,
            away_expected,
            k=K
        )
```
